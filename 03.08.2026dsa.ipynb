 Question 1: Pattern SpottingYou are given a sorted array nums = [1, 2, 4, 6, 10] and a target product K = 24. You must find if two elements multiply to 24 using $O(1)$ extra memory.A) Is this a Single-Array or Double-Array pattern?B) Should the pointers start at Opposite Ends (left = 0, right = len - 1) or Same Direction (slow = 0, fast = 0)?C) If nums[left] * nums[right] > K, which pointer do you move?

```python
nums = [1, 2, 4, 6, 10]
K=  24
def twosum(nums,k):
    left =0
    right = len(nums)-1

    while left<right:
        if nums[right]*nums[left]==k:
            return(nums[right],nums[left])

        elif nums[right]*nums[left]>k:
            right -=1

        else:
            left +=1
            
    return twosum
```

```python
print(twosum(nums,K))
```

 Question 3: Mini Dry RunGiven two sorted arrays:A = [2, 6]B = [1, 3, 5]Using our Merge Step logic (if A[i] < B[j] ... else ...), write down the exact order of elements appended to merged as the while loop runs step-by-step.Take your time! Answer 1, 2, and 3 whenever you are ready.

```python
A=[2,6]
B=[1,3,5]
i=0
j=0                                  #
final =[]
while i <len(A) and j< len(B):
    if A[i]<B[j]:
       final.append(A[i])
       i+=1
    else:
        final.append(B[j])
        j+=1
        
print(final)
final.extend(A[i:])
print(final)
```
