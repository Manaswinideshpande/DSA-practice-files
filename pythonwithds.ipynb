🏋️ Task 5: The NLP Text Token Cleaner
The Scenario: You are building a text preprocessing pipeline for a Machine Learning model (like a spam filter).
Raw text data from the internet is incredibly messy—it contains random spaces, mixed capitalization, 
and sometimes missing values (None). Before feeding this text into a model, you must standardize it.

Your Goal:
Write a defensive function named clean_text_tokens that accepts a list of raw tokens, 
standardizes them to lowercase, strips out unwanted whitespace, removes useless empty strings, and returns a pristine list of cleaned strings.

Your Specific Constraints:

Type Hinting: The function must accept a raw list (tokens: list) and return a clean list of strings (-> list[str]).

Container Placement: Initialize your empty accumulator list named pristine_tokens outside and above your loop.

The Loop & Safety Net: Loop through every item in the input list. Inside a try/except block:

Force the item to a string format using str(item) (this protects the loop if an unexpected None or number is in the text list).

Use string methods to strip out leading/trailing whitespaces (.strip()) and force the text to lowercase (.lower()).

The Filtering Condition: Check if the resulting cleaned string is not empty (!= "").
                                                   If it has actual text, append it to pristine_tokens.
                                                                        If it's just an empty space, let the loop skip it.

Catch exceptions using a broad except Exception: block and print a warning message.

Indentation Check: Ensure your return statement is placed outside the loop so it waits for the entire list to finish processing.

Test Dataset to keep in mind:

Python
raw_tokens = ["  DEEP learning ", " ", "neural_network", None, "  Transformers  ", ""]
# Expected Output: ["deep learning", "neural_network", "none", "transformers"]

```python
from typing import List
def clean_text_tokens(tokens:List[str])->List[str]:
    pristine_tokens =[]
    for data in tokens:
        try:
           item =str(data)
           clean =item.lower().strip()    
           if clean!="":
              pristine_tokens.append(clean)
           else:
              print('skip')
        except Exception as e:
            print("warning text character is missing could not process token {e}")
    return  pristine_tokens
```

```python
raw_tokens = ["  DEEP learning ", " ", "neural_network", None, "  Transformers  ", ""]
```

```python
clean_text_tokens(raw_tokens)
```

Your Goal:Write a function named scale_feature_values that takes a list of raw numeric readings,
checks if they fall within realistic safety bounds, normalizes them, 
and returns a clean list of scaled values.Your Specific Constraints:Type Hinting: Input is a raw list (readings: list).
Output is a list of floats (-> list[float]).Container Placement: Initialize your empty results list called scaled_data outside and above your loop.
The Loop & Safety Net: Loop through every item in the input list. 
Inside a try/except block inside the loop:Cast the item to a float: val = float(item).
The Boundary Condition: Check if the value is strictly between 0.0 and 100.0 (inclusive: $0.0 \le val \le 100.0$).
If it is within bounds, perform a mock scaling math operation: divide the value by 100 (scaled_val = val / 100.0),
and append scaled_val to scaled_data.If it is outside bounds, print "[SKIPPED]: Value out of realistic bounds." and 
let the loop continue.Catch any calculation or type exceptions gracefully with a print statement.
Indentation Check: Ensure your return statement waits for the entire loop to finish
Test Dataset:Pythonraw_readings = [15.0, "missing", 82.4, 9999.0, -45.2, "71.3"]
# Expected Output: [0.15, 0.824, 0.713] (skipping strings, 9999.0, and -45.2)

```python
from typing import List
def scale_feature_values(data:List)->List[float]:
    scaled_data=[]
    for items in data:
        try:
            var = float(items)
            if 0.0 <= var <= 100.0:
                scaled_val = var/100.0
                scaled_data.append(scaled_val)
            else:
                print("[SKIPPED]: Value out of realistic bounds.")
            
            continue
        except Exception as e:
            print("warning unwanted chars and errors found in data could not process the sclaing{e}")
    return scaled_data
    
```

Write a highly defensive function named extract_session_durations that navigates these nested structures safely without crashing or using nested loops.

Type Hinting: Input is a list of dictionaries: logs: list[dict]. Output must return a list of floats: -> list[float].

Container Placement: Initialize an empty list called valid_durations outside and above your loop.

The Loop & Safety Net: Loop through every record dictionary in the logs list. Inside a single try/except block inside the loop:

The Nested Extraction: Access the duration directly by chaining the keys: record["metrics"]["duration"].

Cast that value to a float().

Append it to valid_durations.

The Catch: If either key is missing (KeyError) or the value cannot be cast to a float (ValueError), your except block must catch it,
print "[ERROR]: Log entry corrupt or incomplete.", and let the loop proceed to the next record.

Indentation Check: Ensure your final return statement waits for the entire loop to finish.

Test Dataset to think about:

Pythonapi_logs = [
    {"user": "Alice", "metrics": {"duration": 45.5, "pages": 4}},
    {"user": "Bob", "metrics": {"pages": 2}},               
    {"user": "Charlie"},                                   
    {"user": "David", "metrics": {"duration": "120.0"}},
    {"user": "Eve", "metrics": {"duration": "UNKNOWN"}}
]
# Expected Output: [45.5, 120.0]

```python
from typing import Dict,List,Any
def extract_session_durations(data:List[Dict[str,Any]])->List[float]:
    valid_durations=[]
    for records in data:
        try:
            keys = float(records["metrics"]["duration"])
            valid_durations.append(keys)
        except Exception as e:
            print("value error occured nad cannot process the data {e}")
    return valid_durations
```
