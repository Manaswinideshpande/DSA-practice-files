🔄 Pattern M - Run 2: Median Imputation
The Scenario:
You have a feature column of income data represented as a list of floats, but some entries are missing (None). Because income datasets often contain extreme values, we want to replace None with the median (middle value) of all valid numbers rather than the mean.

Your Goal:
Write a function named impute_missing_with_median that:

Filters out None values and finds the median of the remaining numbers.

Replaces every None in the list with that median value.

Test Case:

Python
incomes = [10.0, 20.0, None, 100.0, None]

1. Valid numbers sorted: [10.0, 20.0, 100.0] -> Median = 20.0
Expected Output: [10.0, 20.0, 20.0, 100.0, 20.0]

```python
#incomes = [10.0, 20.0, None, 100.0, None]
import random
from typing import List
def impute_missing_with_median(data:List)->List[float]:
    filtered=[]
    sortedval = sorted([x for x in data if x is not None])
    if not sortedval:
        return[]
    n=len(sortedval)
    midvalue=n//2
    if midvalue%2 !=0:
        mid_value =sortedval[midvalue]
    else:
        mid_value=sortedval[midvalue-1]+ sortedval[midvalue]/2.0
    for i in data:
        if i is None:
            filtered.append(mid_value)
        else:
            filtered.append(i)
    return filtered

```

```python
incomes = [10.0, 20.0, None, 100.0, None]
result=impute_missing_with_median(incomes)
result
```

Pattern M - Run 3: Mode Imputation (Locking It In)
For categorical/text features (like city names or user status), mean/median doesn't make sense. Instead, we use the Mode (most frequently occurring value).

The Scenario:
You have a feature column of user cities represented as strings, with missing values as None.

Write a function impute_missing_with_mode that:

Finds the most frequent city name (ignoring None).

Replaces all None entries with that mode value.

Test Case:

```python
cities = ["NYC", "London", None, "NYC", "Paris", None]
from typing import List
def impute_missing_with_mode(cities:List[str])->List[str]:
    freq={}
    for i in cities:
        if i in freq:
            freq[i] +=1
        else:
            freq[i] =1
    print(freq)
    mode=max(freq,key=freq.get)
    final=[]
    for i in cities:
        if i is None:
            final.append(mode)
        else:
            final.append(i)
    return final
        
```

```python
print(impute_missing_with_mode(cities))
```
