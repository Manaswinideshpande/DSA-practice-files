Problem Statement
Given an array of integers nums, write a function that returns the most frequent even element.

If there is a tie for the most frequent even element, return the smallest one.

If there are no even elements in the array at all, return -1.

Example Test Cases
Example 1:

Input: nums = [0, 1, 2, 2, 4, 4, 1]

Output: 2

Explanation: Both 2 and 4 appear twice (max frequency of 2). 2 is smaller than 4, so we return 2.

Example 2:

Input: nums = [4, 4, 4, 9, 2, 4]

Output: 4

Explanation: 4 appears 4 times, which is the highest frequency.

Example 3:

Input: nums = [29, 47, 21]

Output: -1

Explanation: No even numbers exist in the list.
