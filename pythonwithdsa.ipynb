```python
from typing import Dict
import requests

def fetch_user_profile(url: str) -> dict:
    try:
        # 1. Clean network request (No extra brackets needed)
        response = requests.get(url)
        
        # 2. Automatically crashes and goes to 'except' if the URL is broken/404
        response.raise_for_status()
        if "application/json" not in response.headers.get("Content-Type", ""):
            print("[WARNING]: Server responded, but it was not JSON data!")
            return {"status": "error", "message": "Invalid Data Format"}
        # 3. If it didn't crash above, it means it's a success! Parse and return immediately
        data = response.json()
        return data
        
    except Exception as e:
        print(f"[WARNING]: URL connection failed. Details: {e}")
        # 4. Explicitly return your fallback dictionary if anything went wrong
        return {"status": "error", "message": "API Offline"}
```
