1. Your Task: The "Production Performance" Analyzer
Imagine you are analyzing the daily output of a factory. Some days the sensors fail (None), and some days the factory is closed (output is 0).

Your Goal: Calculate the average production only for the days the factory was actually running.

Requirements:

Take input using input().split().

Clean: Use try/except to skip typos.

Clean: Skip "None" values.

Filter: Only include numbers greater than 0 (because we only want "active" days).

Analyze: Calculate and print the Average Production.

Analyze: Print the Total Number of Active Days found.

Test Input to try: 100 150 None -10 abc 200 0

```python
test = [100,150,"None",-10,200,0]
count =0
total_average = 0
for item in test:
    if item == "None":
        continue
    try:
        val = float(item)
        if val >0:
            total_average +=val
            count +=1
    except ValueError:
        print(f"Skipping non numeric data:{item}")
if count>0:
    average = total_average/count
    print(f"Total Number of Active Days: {count}")
    print(f"the average production = {average}")
else:
    print("No active days found")
        
    
```

2. Your Task: The "Quality Control" Counter
You are checking items on a conveyor belt.

Weights between 45 and 55 are "Perfect."

Anything else is "Rejected."

Your Goal:

Take input of weights: input().split().

Clean: Skip "None" and use try/except for typos.

Logic:

If the weight is between 45 and 55 (inclusive), add 1 to a perfect_count.

Otherwise, add 1 to a reject_count.

Final Output: Print the total number of Perfect items and the total number of Rejected items.

Test Input: 50 42 None 55 60 48 abc
(Expected result: Perfect: 3, Rejected: 2)

```python
weights1 = input().split()
clean = 0
perfect_count =0
reject_count = 0
for i in weights1:
    if i == "None":
        continue
    try:
        val = int(i)
        if val>=45 and val<=55:
            perfect_count +=1
        else:
            reject_count +=1
    except ValueError:
        print(f"Invalid items : {perfect_count}")
    
print([perfect_count,reject_count])
```
