📝 Task 1: Dry-Run Challenge
Trace the two-pointer execution where pointer j stretches ahead of pointer i to produce a distance greater than 0.

Array A (descending): [10, 8, 4]

Array B (descending): [12, 9, 9]

⚙️ Algorithm Rules:
Start i = 0, j = 0, max_dist = 0

While i < len(A) and j < len(B):

If B[j] >= A[i]:

If j >= i: max_dist = max(max_dist, j - i)

j += 1

Else:

i += 1

```python


def maximumdistance(A: list[int], B: list[int]) -> list[int]:
    i=0
    j=0        #algorithm
    max_distance=0
    while i<len(A) and j<len(B):
        if B[j] >=A[i]:# found valid pair
            if j>=i:
                max_distance=max(max_distance,j-i)
                
            j+=1 # moves j to right if b is greater than a
        else:
            i +=1# moves i to right if b is less than a
    return max_distance

```

```python

A = [10, 8, 4]

B = [12, 9, 9]
result=maximumdistance(A,B)
result
```

💡 Pointer Movement Hints:
Initialize i = 0 and j = 0, plus an empty list result = [].

While i < len(A) and j < len(B):

If A[i] == B[j]: Found a common element! Append to result, then advance both: i += 1, j += 1.

If A[i] < B[j]: A[i] is too small, so move i += 1.

If A[i] > B[j]: B[j] is too small, so move j += 1.

```python
def find_intersection(A: list[int], B: list[int]) -> list[int]:
    i=0
    j=0
    result=[]
    while i<len(A) and j <len(B):
        if A[i]==B[j]:
            result.append(A[i])
            i +=1
            j +=1
        elif A[i]<B[j]:# A is small so move i to right 
            i+=1
        else:
            j+=1
    return result


# Test case:
# A = [1, 2, 4, 5, 6]
# B = [2, 3, 5, 7]
# Expected Output: [2, 5]
```
