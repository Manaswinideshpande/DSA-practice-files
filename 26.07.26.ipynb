Write a function named scale_feature_data:

Type Hints & Signature: Takes data: List[float], min_val: float, and max_val: float, and returns List[float].

Setup: Initialize scaled_data = [].

Loop & Math Logic: Loop through data. Inside a try block:

Apply the Min-Max formula: (val - min_val) / (max_val - min_val)

Round the result to 2 decimal places using round(result, 2).

Append the scaled float to scaled_data.

Guardrail: Inside except Exception as e:, print an error log and append 0.0 as a safe fallback.

Return: Return scaled_data.

Test Dataset:

Python
raw_scores = [10.0, 20.0, 50.0, 100.0]
min_val = 10.0, max_val = 100.0

```python
raw_scores = [10.0, 20.0, 50.0, 100.0]
# min_val = 10.0, max_val = 100.0
from typing import List
def scale_feature_data(data):
    scaled_data = []
    min_val=min(data)
    max_val=max(data)
    value = max_val -min_val
    if value ==0:
        return [0.0]*len(data)
       
    for val in data:
       try:
            min_max = (val-min_val)/value
            result = round(min_max,2)
            scaled_data.append(result)
       except Exception  as e:
            print(f"handled Exception as {e}")
            scaled_data.append(0.0)
    return scaled_data
    

```

```python
scale_feature_data(raw_scores)
```
