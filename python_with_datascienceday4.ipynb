```python
employees = [
    {"name": "Arun", "rating": 4.5, "salary": 50000},
    {"name": "Briya", "rating": 3.2, "salary": 60000},
    {"name": "Chandan", "rating": 4.8, "salary": 70000},
    {"name": "Divya", "rating": 2.9, "salary": 45000}
]

total_bonus = 0
for employee in employees:
    if employee["rating"]>4.0:
        total_bonus =total_bonus+ employee["salary"] *10/100

print(total_bonus)
```

    {"order_id": 101, "item": "Laptop", "price": 1200, "status": "Delivered"},
    {"order_id": 102, "item": "Phone", "price": 800, "status": "Cancelled"},
    {"order_id": 103, "item": "Monitor", "price": 300, "status": "Delivered"},
    {"order_id": 104, "item": "Headphones", "price": 150, "status": "Pending"},
    {"order_id": 105, "item": "Keyboard", "price": 100, "status": "Delivered"}
]
The Objective: Calculate and print the total revenue generated only from orders that have a status of "Delivered".


```python
orders = [
    {"order_id": 101, "item": "Laptop", "price": 1200, "status": "Delivered"},
    {"order_id": 102, "item": "Phone", "price": 800, "status": "Cancelled"},
    {"order_id": 103, "item": "Monitor", "price": 300, "status": "Delivered"},
    {"order_id": 104, "item": "Headphones", "price": 150, "status": "Pending"},
    {"order_id": 105, "item": "Keyboard", "price": 100, "status": "Delivered"}
]
total_revenue = 0
for order in orders:
    if order['status'] == 'Delivered':
        total_revenue = total_revenue + order['price']
print( total_revenue )
```
