```python
from typing import List

def countDuplicates(nums: list[int]) -> int:
        k = 0  # unique tracker

        for i in range(len(nums)):
            if nums[i] != nums[k]:
                k += 1
                nums[k] = nums[i]  #  Destination (nums[k]) = Source (nums[i])

        return k + 1
            
```

```python
nums=[0,0,0,1]
print(countDuplicates(nums))
```

```python
#nums = [1,1,2]
nums = [0,0,1,1,1,2,2,3,3,4]
```

```python
print(countDuplicates(nums))
```
