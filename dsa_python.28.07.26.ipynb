### Here we use Hash Maps for position tracking ({value: index}).

💡 The Big Concept: "Looking for the Complement"Imagine you have an array nums = [2, 7, 11, 15] and a target = 9. You want to find the two indices whose values add up to 9.Instead of running two nested loops (which takes $O(N^2)$ time), you iterate through the list once ($O(N)$ time). At each step, ask:$$\text{complement} = \text{target} - \text{current\_number}$$"I am at number $X$. Has my complement $(\text{target} - X)$ already been seen in the map?"If YES: You found your pair! Return [seen_map[complement], current_index].If NO: Store the current number and its index in your map: seen_map[X] = current_index.



```python
def twoSum(nums:list[int], target: int) -> list[int]:
    seen = {}  
    for index, num in enumerate(nums):
        complement = target-num
        if complement in seen:
            return[seen[complement],index]
        else:
            seen[num]=index
    return []
        
```

```python
nums = [2, 11, 7, 15]
target = 9
result1=twoSum(nums,target)
result1
```

Given an array of integers nums and a target integer target, write a function find_pair_indices(nums, target) that returns the indices of the two numbers that add up to target.

Example Test Cases

Input: nums = [10, 15, 3, 7], target = 17

Output: [0, 3]

Explanation: nums[0] + nums[3] is 10 + 7 = 17.

Explanation: nums[1] + nums[2] is 5 + 8 = 13.

Output: []

```python
def find_pair_indices(nums: list[int], target: int) -> list[int]:
    seen = {}  # Format: {number: index}
    for index, val in enumerate(nums):
        complement = target - val
        if complement in seen:
            return[seen[complement],index]
        else:
            seen[val] = index
    return []
```
