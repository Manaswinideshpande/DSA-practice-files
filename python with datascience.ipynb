```python
def safe_sum_all(*a)->tuple:
    """ If the user enter is number it sums all the numbers from tuple or 
    list if user entery is string then exception handles the error without disturbing the flow of code"""
    sum_=0
    try:
        for i in a:
            sum_ = sum_+i
        print(sum_)
    except Exception as e:
        print("exception is text return 0.0")
              
```
