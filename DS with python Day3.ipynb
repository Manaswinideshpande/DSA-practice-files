1. The Logic: Counting with DictionariesOne of the most powerful things a Data Scientist does is "Categorization."
Let's say you have a list of fruit, and you want to count how many of each you have.

The Algorithm:
Start with an empty dictionary: counts = {}.Loop through your list.
    
If the item is already in the dictionary,add 1 to its value

If it is not in the dictionary, add it and set the value to $1$.

```python
# To count more than one fruits in a box;Method 1

fruits = input("").split()
dic1 ={}
count = 0
for i in fruits:
    if i in dic1:
        dic1[i] = dic1[i]+1
        count +=1
    else:
        dic1[i] =1
print(dic1)
print(count)
```

```python
#To count total number of all fruits in the box;Method 2
fruits = input("").split()
dic1 ={}
count = 0
for i in fruits:
    count +=1
    if i in dic1:
        dic1[i] = dic1[i]+1
    else:
        dic1[i] =1
print(dic1)
print(count)
```

2. Your Task: The "Social Media Tag" CounterYou are analyzing a list of hashtags from a social media feed.

You need to count how many times each hashtag appears to find the "Trending" topics.

Your Goal:

Take input of tags (space-separated): input().split().

Create an empty dictionary called tag_counts.

Clean: If a tag is just #, skip it (too short).

Logic: Loop through the tags.

If the tag exists in tag_counts, increment its value.

If it doesn't exist, create the key and set it to 1.Final Output: Print the dictionary.

Test Input: #python #data #python #ai #data #python(Expected Output: {'#python': 3, '#data': 2, '#ai': 1})

Why this is useful: This is the first step in Natural Language Processing (NLP). Before an AI can understand text, it has to count the frequency of words!

```python
tags = input().split()
tag_counts={}
for tag in tags:
    if tag == '#':
        continue
    if tag in tag_counts: 
        tag_counts[tag] = tag_counts[tag]+1
    else:
        tag_counts[tag] = 1
print(tag_counts)
```

🛒 The Task: Store Checkout
Use this code as your starting point:

Python
price_list = {
    "apple": 2,
    "banana": 1,
    "orange": 3
}

YOUR TASK: 
1. Loop through the cart.

2. If the item is in price_list, add its price to total_bill.

3. If NOT, print a message saying it's out of stock.

4. Print the final total_bill.

```python
price_list = {
    "apple": 2,
    "banana": 1,
    "orange": 3
}
cart = input("Enter items to buy: ").split()
total_bill = 0

for item in cart:
    if item in price_list:
        price = price_list[item]
        total_bill = total_bill + price
    else: 
        print(f"sorry , we don't have {item} in stock")

print(f"Your final total is: ${total_bill}")
```
