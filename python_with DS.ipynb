```python
def extract_clean_ages (data)->list[int]:
    valid_ages =[]
    for record in data:
         for key, values in record.items():
             if key =='age':
                 try:
                    valid_ages.append(int(values))
                 except Exception as e:
                    print("[SKIPPED]: Invalid or missing age encountered.")
    return valid_ages
```

```python
from typing import Any, Dict,List
def extract_clean_ages (data:List[Dict[str,Any]])->list[int]:
    valid_ages =[]
    for record in data:
        try:
            ages = int(record['age'])
            valid_ages.append(ages)
        except Exception as e:
            print("[SKIPPED]: Invalid or missing age encountered.")
    return valid_ages
            
```

```python
from typing import Any,Dict,List
def  filter_high_temperatures(data:List[Dict[str,Any]])->list[float]:
    high_temp_list=[]
    for records in data:
        try:
            temp = float(records["temperature"])
            if temp>=100:
               high_temp_list.append(temp)
            if 'temperature' not in records:
                print("temperature missing exception handles")
        except Exception as e:
            print("[ERROR]: Sensor reading corrupt.")    
            
    return high_temp_list
    
```
