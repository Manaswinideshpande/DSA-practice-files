First Non-Repeating Character
1. The Story & The Pattern
The Scenario:

Imagine guests checking into a hotel. Some guests brought family members (duplicates), and some came alone. The manager wants to award a prize to the very first guest in line who came completely alone.

Input: "swiss"

Output: 'w'

(Explanation: 's' appears 3 times, 'w' appears 1 time, 'i' appears 1 time. Even though 'i' is also unique, 'w' came first in line!)

The Strategy (Two-Pass Method):

Pass 1: Walk down the line from start to finish and build your frequency map (tally sheet).

Pass 2: Walk down the original line a second time. Check your tally sheet for each character. The first one with a count of 1 is your winner!

```python
n="swiss"
#o='w'
guest={}
for i in n:
   if i not in guest:
      guest[i] =1
   else:
      guest[i] +=1
for j in n:
   if guest[j]==1:
      result = j
      break
print(result)
```
