📋 Your Goal & Constraints (Task 28 - Outlier Capping)
Write a function named cap_outliers:

Type Hints & Signature: Takes data: List[float], q3: float, and iqr: float, returning List[float].

Setup:

Calculate upper limit: upper_bound = q3 + (1.5 * iqr)

Initialize capped_data = []

Loop Logic (Inside try block inside the loop):

If val > upper_bound, cap it by appending upper_bound to capped_data.

Else, append val directly.

Guardrail: Catch exceptions in an except Exception as e: block, log it, and append 0.0.

Return: Return capped_data.

Test Dataset:

Python
purchases = [25.0, 45.0, 50.0, 60.0, 85.0, 1000.0]  # 1000.0 is an extreme outlier
Given parameters: q3 = 72.5, iqr = 40.0
Upper bound will be: 72.5 + (1.5 * 40.0) = 132.5

```python
from typing import List
def cap_outliers(data,q3,iqr)->List[float]: 
    capped_data =[]
    for val in data:
        try:
            upper_limit = q3+(1.5*iqr)
            if val>upper_limit:
                capped_data.append(upper_limit)
            else:
                capped_data.append(val)
        except Exception as e:
            print(f"Exception error handled as {e}")
            capped_data.append(0.0)
    return capped_data
```

```python
purchases = [25.0, 45.0, 50.0, 60.0, 85.0, 1000.0]
result = cap_outliers(purchases, q3=72.35,iqr=40.0)
result
```

📋 Your Goal & Constraints (Task 29 - Outlier Trimming)
Write a function named trim_outliers:

Type Hints & Signature: Takes data: List[float], q3: float, and iqr: float, returning List[float].

Setup:

Calculate upper_bound = q3 + (1.5 * iqr)

Initialize trimmed_data = [] outside the loop.

Loop Logic (Inside try block inside loop):

If val <= upper_bound (it is NOT an outlier), append val to trimmed_data.

Else (if it's an outlier), do nothing! (Let it drop).

Guardrail: Catch exceptions with except Exception as e:, log it, and do not append bad data.

Return: Return trimmed_data.

Test Dataset:

Python
house_prices = [150000.0, 200000.0, 250000.0, 1200000.0]  # 1.2M is an outlier
Parameters: q3 = 250000.0, iqr = 100000.0
Upper bound will be: 250000.0 + (1.5 * 100000.0) = 400000.0

```python
def trim_outliers(data,q3,iqr)->List[float]:
    trimmed_data = []
    upper_bound = q3 + (1.5 * iqr)
    for i in data:
        try:
            if i<=upper_bound:
                trimmed_data.append(i)
            else:
                print("drop it")
        except Exception as e:
            print(f"Exception error handled as {e}") 
            trimmed_data.append(0.0)
    return trimmed_data
        


```
