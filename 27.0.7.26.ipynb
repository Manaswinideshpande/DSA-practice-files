Write a function named cap_outliers:

Type Hints & Signature: Takes data: List[float], q3: float, and iqr: float, returning List[float].

Setup:

Calculate upper limit: upper_bound = q3 + (1.5 * iqr)

Initialize capped_data = []

Loop Logic (Inside try block inside the loop):

If val > upper_bound, cap it by appending upper_bound to capped_data.

Else, append val directly.

Guardrail: Catch exceptions in an except Exception as e: block, log it, and append 0.0.

Return: Return capped_data.

Test Dataset:

Python
purchases = [25.0, 45.0, 50.0, 60.0, 85.0, 1000.0]  # 1000.0 is an extreme outlier
Given parameters: q3 = 72.5, iqr = 40.0
Upper bound will be: 72.5 + (1.5 * 40.0) = 132.5

```python
from typing import List
def cap_outliers(data,q3,iqr)->List[float]: 
    capped_data =[]
    for val in data:
        try:
            upper_limit = q3+(1.5*iqr)
            if val>upper_limit:
                capped_data.append(upper_limit)
            else:
                capped_data.append(val)
        except Exception as e:
            print(f"Exception error handled as {e}")
            capped_data.append(0.0)
    return capped_data
```
