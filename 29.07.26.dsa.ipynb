Two-Pointer Technique (Sorted Arrays)
💡 The Big IdeaWhen an array is already sorted in ascending order, we can find pair sums in $O(N)$ time using $O(1)$ extra space (zero extra memory) without needing a Hash Map!🛠️
 How Two Pointers WorkWe place two pointers at opposite ends of the array:left = 0 (pointing to the smallest element)right = len(nums) - 1 (pointing to the largest element)At each step, calculate current_sum = nums[left] + nums[right]:
 

```python
nums=[1, 3, 4, 6, 8, 10]
target = 10
from typing import List
def sorted_array(nums:list[int],target:int)->List[int]:
    left = 0
    right=len(nums)-1
    while left<right:
        current_sum = nums[left] + nums[right]
        if current_sum==target:
           return[left,right]
        elif current_sum<target:
            left +=1
        else:
            right -=1
    return[]
```

```python
result=sorted_array(nums,target)
result
```

```python
#Sorted Array: nums = [2, 5, 8, 12, 15]
#Target: 20
from typing import List
def sorteddata(nums:List,target:int)->List[int]:
    left = 0
    right=len(nums)-1
    while left <right:
        current_sum=nums[left]+nums[right]
        if current_sum==target:
            return[left,right]
        elif current_sum<target:
            left +=1
        else:
            right -=1
    return []

```

```python
nums = [2, 5, 8, 12, 15]
target=20
result1=sorteddata(nums,target)
```
