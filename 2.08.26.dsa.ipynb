Problem 1: Find Common Elements (Intersection)
Given two ascending sorted arrays A and B, return a list containing all elements present in both arrays.

Array A: [2, 5, 8, 11, 15]

Array B: [1, 5, 9, 11, 12, 16]

Expected Output: [5, 11]

```python
A= [2, 5, 8, 11, 15]
B= [1, 5, 9, 11, 12, 16]
i=0
j=0
a=[]
while i<len(A) and j<len(B): #j=1,i=2 
    if B[j] == A[i]:
       a.append(A[i])
       j +=1
       i +=1
    elif A[i]<B[j]:
        i +=1
    else:
        j +=1
       
print(a)
    



        
```

```python
A= [1, 3, 5, 8]
B =[2, 4, 6, 7, 9, 10]
i=0
j=0
merged =[]
while i<len(A) and j<len(B):#1<2=1,
    if A[i]<B[j]:
     
      merged.append(A[i])
      i+=1
    else:
      merged.append(B[j])
      j+=1

merged.extend(A[i:])
merged.extend(B[j:])
print(merged)
```
