📋 Question 1: Pattern Recognition (Spatial Intuition)
You are given a sorted array containing negative and positive numbers, like [-7, -3, 1, 4, 8]. You need to create an array of their squares, also in sorted order [1, 9, 16, 49, 64].

A) Which of the 3 sub-patterns does this belong to? (Same-Direction Reader/Writer, Opposite-Ends Convergence, or Back-to-Front Buffer Fill)

B) Where should your two pointers start, and why?

```python
arr=[-7, -3, 1, 4,8] # sorted 1 array, opposite-ends pointers from 0 to len(arr) ,use left and right poniters since it needs sorting the array
left=0
right = len(arr)-1
new=[]
while left<=right:
    left_square= arr[left]**2
    right_square=arr[right]**2
    if left_square > right_square:
       new.append(left_square)
       left +=1
    else:
       new.append(right_square)
       right -=1
        
new.reverse()
print(new)
```
